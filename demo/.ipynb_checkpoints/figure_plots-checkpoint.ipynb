```python
import os
import re
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from scipy.stats import ttest_ind
import numpy as np
from pathlib import Path
```

```python
# -----------------------------
# 1) 解析 GCN 文本结果为 DataFrame
# -----------------------------
def parse_gcn_result_txt(txt_path,
                         method_name="gcn",
                         source_name="AResKG",   # 你也可以改成 "OURS" 或其它
                         keep_metrics=("precision", "recall", "f1", "map")):
    """
    读取类似于 res_gcn.txt 的块状结果，返回长表：
    columns = [Method, Fold, Metric, Score, Source]
    """
    txt = Path(txt_path).read_text(encoding="utf-8", errors="ignore")

    # 按空行切块：每块就是一个 fold 的结果
    blocks = [b.strip() for b in re.split(r"\n\s*\n", txt) if b.strip()]
    rows = []

    for i, block in enumerate(blocks):
        fold = f"fold_{i}"
        # 抓取 key: value 行
        for line in block.splitlines():
            m = re.match(r"^\s*([A-Za-z0-9_]+)\s*:\s*([0-9.eE+-]+)", line.strip())
            if not m:
                continue
            key, val = m.group(1).lower(), float(m.group(2))

            # 统一指标命名：ap -> map
            #if key == "ap":
                #metric = "map"
            if key == "pr_auc":
                metric = "auc_pr"
            else:
                metric = key

            # 只保留与 df_kids 对齐的四项（可按需扩展）
            if metric in keep_metrics:
                rows.append({
                    "Method": method_name,
                    "Fold": fold,
                    "Metric": metric,
                    "Score": val,
                    "Source": source_name
                })

    df = pd.DataFrame(rows, columns=["Method", "Fold", "Metric", "Score", "Source"])
    return df

def extract_predict_results(data_path, strain, methods):
    data = []

    for method in methods:
        for fold in folds:
            if method == 'pra':
                result_file = os.path.join(data_path, method, "model_instance", fold, "instance",  "test", "results", "results.txt")
            else:
                result_file = os.path.join(data_path, method, "model_instance", fold, "test", "results", "results.txt")
            if not os.path.exists(result_file):
                print(f"Missing: {result_file}")
                continue
                
            with open(result_file, "r") as f:
                content = f.read()
        
            # 提取四个指标
            results = {}
            for metric in metrics_of_interest:
                match = re.search(fr"{metric}:\s*([0-9.]+)", content)
                if match:
                    results[metric] = float(match.group(1))
                else:
                    results[metric] = None
        
            for metric, value in results.items():
                data.append({
                    "Method": method,
                    "Fold": fold,
                    "Metric": metric,
                    "Score": value,
                    "Source": strain
                })
    return pd.DataFrame(data)

def significance_stars(p): 
    if p <= 1e-4: 
        return "****" 
    elif p <= 1e-3: 
        return "***" 
    elif p <= 1e-2: 
        return "**" 
    elif p <= 0.05: 
        return "*" 
    return None
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np

def plot_strain_subplots_metrics_x(
    df_results: pd.DataFrame,
    layout: str = "1x5",
    top_n_strains: int = 5,
    strain_order: list = None,
    #metrics_order = ("Precision","Recall","F1","Average Precision"),
    metrics_order = ("Precision","Recall","F1","AUCPR"),
    method_order  = ("PRA","MLP","Stacked","GCN","GAT"),
    palette = ("#6BAED6","#FD8D3C","#74C476","#9E9AC8","#FFB74D"),
    ylim=(0,1.0),
    add_points: bool = True,
    savepath: str = None
):
    """绘制多个菌株子图：x=Metric, y=Performance, hue=Method（箱型图+可选散点）"""
    need = {"Method","Metric","Score","Source"}
    if not need.issubset(df_results.columns):
        raise ValueError(f"df_results must contain columns: {need}")

    df = df_results.copy()
    df["Metric"] = df["Metric"].str.lower()
    metrics_order = [m.lower() for m in metrics_order]
    df = df[df["Metric"].isin(metrics_order)]
    df["Metric"] = pd.Categorical(df["Metric"], categories=metrics_order, ordered=True)

    # 选取 topN 菌株
    if strain_order is None:
        strain_order = df["Source"].value_counts().head(top_n_strains).index.tolist()
    else:
        strain_order = [s for s in strain_order if s in df["Source"].unique()]
    df = df[df["Source"].isin(strain_order)]

    # 方法顺序与配色
    method_order = [m for m in method_order if m in df["Method"].unique()]
    pal = dict(zip(method_order, palette[:len(method_order)]))

    # 布局设置
    if layout == "1x5":
        nrows, ncols = 1, 5
    elif layout == "2x3":
        nrows, ncols = 2, 3
    else:
        raise ValueError("layout must be '1x5' or '2x3'.")

    sns.set(style="whitegrid", font_scale=1.25)
    fig, axes = plt.subplots(
        nrows, ncols,
        figsize=(8*ncols, 5*nrows),
        sharey=True
    )
    axes = np.array(axes).flatten()

    # 绘制每个菌株
    for i, strain in enumerate(strain_order):
        ax = axes[i]
        dat = df[df["Source"] == strain]

        sns.boxplot(
            data=dat,
            x="Metric", y="Score",
            hue="Method",
            hue_order=method_order,
            order=metrics_order,
            palette=pal,
            ax=ax
        )

        if add_points:
            sns.stripplot(
                data=dat,
                x="Metric", y="Score",
                hue="Method",
                hue_order=method_order,
                order=metrics_order,
                dodge=False,
                palette=["#000000"],  # 避免 FutureWarning
                size=3, alpha=0.6,
                ax=ax
            )

        # 清理多余图例（仅第一层箱图会产生 legend）
        if ax.legend_ is not None:
            ax.legend_.remove()

        ax.set_title(strain, fontsize=14, pad=8)
        ax.set_ylim(*ylim)
        ax.set_ylabel("Performance", fontsize=16)
        ax.set_xlabel("")              # 不显示x轴标题
        ax.set_title(f"$\\it{{{strain}}}$", fontsize=16, pad=8)
        ax.set_xticklabels(["Precision","Recall","F1","AUCPR"])

    # 隐藏空子图（如 2x3）
    for j in range(len(strain_order), len(axes)):
        axes[j].axis("off")

    # 添加统一图例（右侧居中）
    handles, labels = axes[0].get_legend_handles_labels()
    fig.legend(handles[:len(method_order)], labels[:len(method_order)],
               title="Method", loc="center left", bbox_to_anchor=(0.7, 0.3), frameon=False)

    plt.subplots_adjust(wspace=0.15, hspace=0.25, right=0.85)
    #plt.suptitle(f"Model Performance Across Metrics (per Strain) — layout {layout}", fontsize=16, y=1.02)

    if savepath:
        plt.savefig(savepath, dpi=300, bbox_inches="tight")
    plt.show()


# 1x5 布局（常见于 Supplementary Figure）
#plot_strain_subplots_metrics_x(df_results, layout="1x5")

# 2x3 布局（适合主文 Figure）
plot_strain_subplots_metrics_x(
    df_results,
    layout="2x3",
    strain_order=["Staphylococcus", "Escherichia", "Pseudomonas", "Mycobacterium", "Klebsiella"],
    add_points=False,
    ylim=(0, 1.05),
    savepath="/opt/ai4g_chriszyyang/buddy1/2_project_ongoing/4-antibio_resistance/LLM_ResistantDataExtractor/figure_v1/Figure4_AResKG_top5_species_boxplot.pdf"
)
```

```python
def parse_gcn_result_txt(txt_path,
                         method_name="gcn",
                         source_name="AResKG",   # 你也可以改成 "OURS" 或其它
                         keep_metrics=("precision", "recall", "f1", "map")):
    """
    读取类似于 res_gcn.txt 的块状结果，返回长表：
    columns = [Method, Fold, Metric, Score, Source]
    """
    txt = Path(txt_path).read_text(encoding="utf-8", errors="ignore")

    # 按空行切块：每块就是一个 fold 的结果
    blocks = [b.strip() for b in re.split(r"\n\s*\n", txt) if b.strip()]
    rows = []

    for i, block in enumerate(blocks):
        fold = f"fold_{i}"
        # 抓取 key: value 行
        for line in block.splitlines():
            m = re.match(r"^\s*([A-Za-z0-9_]+)\s*:\s*([0-9.eE+-]+)", line.strip())
            if not m:
                continue
            key, val = m.group(1).lower(), float(m.group(2))

            # 统一指标命名：ap -> map
            if key == "pr_auc":
                metric = "map"
            #if key == "pr_auc":
            #    metric = "auc_pr"
            else:
                metric = key

            # 只保留与 df_kids 对齐的四项（可按需扩展）
            if metric in keep_metrics:
                rows.append({
                    "Method": method_name,
                    "Fold": fold,
                    "Metric": metric,
                    "Score": val,
                    "Source": source_name
                })

    df = pd.DataFrame(rows, columns=["Method", "Fold", "Metric", "Score", "Source"])
    return df

def extract_predict_results(data_path, methods):
    data = []

    for method in methods:
        for fold in folds:
            if method == 'pra':
                result_file = os.path.join(data_path, method, "model_instance", fold, "instance",  "test", "results", "results.txt")
            else:
                result_file = os.path.join(data_path, method, "model_instance", fold, "test", "results", "results.txt")
            if not os.path.exists(result_file):
                print(f"Missing: {result_file}")
                continue
                
            with open(result_file, "r") as f:
                content = f.read()
        
            # 提取四个指标
            results = {}
            for metric in metrics_of_interest:
                match = re.search(fr"{metric}:\s*([0-9.]+)", content)
                if match:
                    results[metric] = float(match.group(1))
                else:
                    results[metric] = None
        
            for metric, value in results.items():
                data.append({
                    "Method": method,
                    "Fold": fold,
                    "Metric": metric,
                    "Score": value
                })
    return pd.DataFrame(data)

def significance_stars(p): 
    if p <= 1e-4: 
        return "****" 
    elif p <= 1e-3: 
        return "***" 
    elif p <= 1e-2: 
        return "**" 
    elif p <= 0.05: 
        return "*" 
    return None
```

```python
name_map = {
    "precision": "Precision",
    "recall": "Recall",
    "f1": "F1",
    "map": "AUCPR"
}

method_map ={
    "pra": "PRA",
    "er_mlp": "MLP",
    "stacked": "Stacked",
    "gat": "GAT",
    "gcn": "GCN"
}
df_long = (df_compare.melt(id_vars=["Method", "Fold", "Metric"], value_vars=["KIDS", "AResKG_Ecoli"], var_name="Source", value_name="Score"))
df_long["Metric"] = (df_long["Metric"].astype(str).str.strip().str.lower().map(name_map))
df_long["Method"] = (df_long["Method"].astype(str).str.strip().str.lower().map(method_map))
```

```python
# ====== 固定方法与指标顺序 ======
method_order = ["PRA", "MLP", "Stacked", "GCN", "GAT"]
metric_order = ["Precision", "Recall", "F1", "AUCPR"]

# 转换为有序分类类型，确保绘图顺序固定
df_long["Method"] = pd.Categorical(df_long["Method"], categories=method_order, ordered=True)
df_long["Metric"] = pd.Categorical(df_long["Metric"], categories=metric_order, ordered=True)
```

```python
### 
methods = [m for m in method_order if (df_long["Method"] == m).any()]
n_methods = len(methods)

sns.set(style="whitegrid", font_scale=1.2)
#fig, axes = plt.subplots(1, n_methods, figsize=(6 * n_methods, 5), sharey=True)
fig, axes = plt.subplots(n_methods, 1, figsize=(10, 3 * n_methods), sharex=True, sharey=True)


if n_methods == 1:
    axes = [axes]  # 确保可迭代

width = 0.36
metrics_order = ["Precision", "Recall", "F1", "AUCPR"]

for ax, method in zip(axes, methods):
    dfg = df_long[df_long["Method"] == method]

    # === 均值 ===
    mean_tbl = (
        dfg.groupby(["Metric", "Source"])["Score"]
        .mean().unstack("Source")
        .reindex(metrics_order)
    )
    for col in ["KIDS", "AResKG_Ecoli"]:
        if col not in mean_tbl.columns:
            mean_tbl[col] = np.nan

    x = np.arange(len(metrics_order))
    bars1 = ax.bar(x - width/2, mean_tbl["KIDS"].values, width, label="KIDS")
    bars2 = ax.bar(x + width/2, mean_tbl["AResKG_Ecoli"].values, width, label="AResKG_Ecoli")

    # === 柱顶数值 ===
    for bars in (bars1, bars2):
        for b in bars:
            h = b.get_height()
            if np.isnan(h):
                continue
            ax.text(b.get_x() + b.get_width()/2, h + 0.01, f"{h:.2f}",
                    ha="center", va="bottom", fontsize=9)

    # === 显著性标注 ===
    y_max = np.nanmax([b.get_height() for b in list(bars1) + list(bars2)])
    if np.isnan(y_max): y_max = 0.0
    y_start_base = y_max + 0.02
    h = 0.01

    for i, metric in enumerate(metrics_order):
        sub = dfg[dfg["Metric"] == metric]
        g1 = sub[sub["Source"] == "KIDS"]["Score"].dropna()
        g2 = sub[sub["Source"] == "AResKG_Ecoli"]["Score"].dropna()
        if len(g1) >= 2 and len(g2) >= 2:
            stat, pval = ttest_ind(g1, g2, equal_var=False)
            sig = significance_stars(pval)
            if sig:
                x1 = x[i] - width/2
                x2 = x[i] + width/2
                y_pair_max = np.nanmax([mean_tbl.iloc[i]["KIDS"], mean_tbl.iloc[i]["AResKG_Ecoli"]])
                y0 = max(y_start_base, (y_pair_max if not np.isnan(y_pair_max) else 0) + 0.02)
                ax.plot([x1, x1, x2, x2], [y0, y0 + h, y0 + h, y0], lw=1.2, c='k')
                ax.text((x1 + x2) / 2, y0 + h + 0.005, sig, ha='center', va='bottom', color='k')

    # === 轴与标题 ===
    ax.set_xticks(x)
    ax.set_xticklabels(metrics_order)
    ax.set_ylim(0, 1.05)
    ax.set_title(method)
    ax.grid(axis="y", linestyle="--", alpha=0.3)

axes[0].set_ylabel("Score")
axes[-1].legend(frameon=False, loc="upper right")
#fig.suptitle("Performance (KIDS vs AResKG_Ecoli)", fontsize=16, y=1.02)
plt.tight_layout()
plt.savefig(f"{home_path}/figure_v1/Figure4_bar_plot_compare.pdf", dpi=300, bbox_inches="tight")
plt.show()
```
